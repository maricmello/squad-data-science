# 02 - Pré-processamento

Com os dados já conhecidos (notebook 01), a etapa aqui é preparar tudo para a clusterização. O K-Means mede distância entre países, então a escala de cada variável importa: se uma coluna varia na casa dos milhares (renda) e outra varia entre 1 e 8 (fertilidade), a de maior escala domina a distância sozinha, mesmo que não seja mais importante que as outras.

A solução padrão para isso é a padronização (z-score): transformar cada variável para ter média 0 e desvio padrão 1.

## Importando bibliotecas

```python
import sys
sys.path.append('..')

import pandas as pd
import joblib

from src.preprocessing import TODAS_AS_FEATURES, dataframe_padronizado
```

## Carregando Dados

```python
df = pd.read_csv('../data/raw/dataset_paises.csv')
df.head()
```

## Padronizando as variáveis

```python
df_padronizado, scaler = dataframe_padronizado(df, TODAS_AS_FEATURES)
df_padronizado.describe().round(2).T[['mean', 'std', 'min', 'max']]
```

## Salvando os dados processados

```python
df_padronizado.to_csv('../data/processed/paises_padronizados_todas_features.csv', index=False)
joblib.dump(scaler, '../data/processed/scaler_todas_features.joblib')

print("Arquivos salvos em data/processed/:")
print("- paises_padronizados_todas_features.csv")
print("- scaler_todas_features.joblib")
```
